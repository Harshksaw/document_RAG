```python
import logging
import os
from datetime import datetime
```

```python
logs_dir = os.path.join(os.getcwd(), "logs")
```

```python
os.makedirs("logs", exist_ok=True)
```

```python
LOG_FILE = os.path.join(logs_dir, f"experiment_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log")
```

```python
LOG_FILE
```

```python
logging.basicConfig(
    filename=LOG_FILE,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
)
```

```python
logger = logging.getLogger("DocumentPortal")
logger.info("Logging is set up.")
```
